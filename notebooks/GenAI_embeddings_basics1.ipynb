```python
text = ['Bright birds greeted dawn',
        'The curious fox vanished quickly',
        'Heavy rain soaked empty streets',
        'Children laughed beneath tall trees today',
        'Fresh coffee filled every room']
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer()
cv.fit_transform(text)
```

```python
cv.get_feature_names_out()
```

```python
cv.fit_transform(text).toarray()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tf = TfidfVectorizer(ngram_range = (1,2))
tf.fit_transform(text)
```

```python
tf.get_feature_names_out()
```

```python
import pandas as pd

df = pd.DataFrame(tf.fit_transform(text).toarray(), columns=tf.get_feature_names_out())
df
```

```python
import pandas as pd
```

```python
columns = ['target','id','date','flag','user','text']
```

```python
path = 'file_path'

sentiment_data = pd.read_csv(path, header=None, names=columns, nrows=1000, encoding='latin-1')
sentiment_data.head()
```

```python
text_data = sentiment_data[['text']]
text_data.head()
```

```python
model_miniLM = SentenceTransformer("all-MiniLM-l6-v2")
emb_miniLM = model_miniLM.encode(text_data['text'].to_list())
emb_miniLM.shape
```
