```python
for i, item in enumerate(retail_queries, start=1):
    item["id"] = f"query_{i}"
```

```python
retail_queries
```

```python
vectors = []

for query in retail_queries:

    embedding = model.encode(query["query"]).tolist()

    vectors.append({
        "id": query["id"], # Added unique ID for each vector
        "values": embedding,
        "metadata": {
            "cat": query["category"],
            "attributes": query["attributes"]
        }
    })
```

```python
index.upsert(vectors=vectors)
```

```python
user_query = "I need a lightweight laptop for programming"
```

```python

user_query_embedding = model.encode(user_query).tolist()
```

```python
results = index.query(
    vector=user_query_embedding,
    top_k=3,
    include_metadata=True
)
```

```python
results
```

```python
for match in results["matches"]:
    print(match)
```

```python
for i, match in enumerate(results["matches"], start=1):
    print(f"Rank {i}")
    print(f"ID:         {match['id']}")
    print(f"Score:      {match['score']:.4f}")
    print(f"Category:   {match['metadata']['cat']}")
    print(f"Attributes: {match['metadata']['attributes']}")
    print("-" * 50)
```
